```python
# python std library
import csv

# analysis and visualization libraries installed from requirements.txt
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data_path = 'data/products-100.csv'
data_path_out = 'data/products-100-out.csv'
```

```python
### open with standard library csv module
# open the file defined at data_path
every_other_row = []
with open(data_path, 'r') as file:
    # use the reader method to load the file as a csv
    reader = csv.reader(file)
    # print out headers
    headers = next(reader)
    print(headers)
    # append rows
    for idx, row in enumerate(reader):
        # if idx is divisible by 2 append it to our list to write out
        if idx % 2 == 0:
            every_other_row.append(row)
```

```python
with open(data_path_out, 'w') as file:
    writer = csv.writer(file)
    writer.writerow(headers)
    writer.writerows(every_other_row)
```
